```python
%pip install pandas
%pip install bertopic
import pandas as pd
import numpy as np
from bertopic import BERTopic
df = pd.read_csv('guardian_articles.csv')
print(f"DataFrame shape: {df.shape}")
mask = (
    (df['sectionName'].isin(['Politics', 'Opinion']))
)

titles = df.loc[mask, 'webTitle'].dropna().tolist()
titles = [t.strip() for t in titles if len(t.split()) > 3]
print(f"Number of titles selected: {len(titles)}")

topic_model = BERTopic(embedding_model="all-MiniLM-L6-v2")
```

```python
topic, probs = topic_model.fit_transform(titles)
```

```python
topic_model.get_topic_info()
```

```python
for i in range(20):
    print(topic_model.get_topic(i))
```
