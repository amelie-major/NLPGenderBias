Then load the CSV file into a dataframe

```python
df = pd.read_csv('/Users/ameliemajor/Desktop/NLPGenderBias/data/guardian_articles.csv')

def clean_html_and_whitespace(text):
    # Protect against non-string (e.g. NaN) values
    if not isinstance(text, str):
        return ''
    text = re.sub(r'<[^>]+>', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

# Apply cleaning and store back into the dataframe
df = df.loc[df['sectionName'] == "Politics"]
df['bodyContent'] = df['bodyContent'].apply(clean_html_and_whitespace)

# Join all rows into one long string
all_text = "\n\n".join(df['bodyContent'].tolist())

# Write to text file
with open("combined_text.txt", "w", encoding="utf-8") as f:
    f.write(all_text)

print("Saved guardian_combined_text.txt")
```

Here, I am going to analyze tweets. First, I need break down the text into

```python
!pip install emoji
!pip install contractions
import re
import pandas as pd
import emoji
from nltk.tokenize import TweetTokenizer
import contractions
def clean_tweet(text):
    # Convert to string (safety)
    text = str(text)

    # Expand contractions (she’s -> she is)
    text = contractions.fix(text)

    # Lowercase
    text = text.lower()

    # Replace URLs
    text = re.sub(r'http\S+|www.\S+', '<URL>', text)

    # Replace user mentions
    text = re.sub(r'@\w+', '<USER>', text)

    # Replace RT
    text = re.sub(r'\brt\b', '<RETWEET>', text)

    # Remove emoji but keep text markers, or convert emoji → text
    text = emoji.replace_emoji(text, replace='')

    # Remove punctuation (but allow ! and ? because sentiment value)
    text = re.sub(r'[^\w\s!?]', '', text)

    # Remove extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()

    return text
```
